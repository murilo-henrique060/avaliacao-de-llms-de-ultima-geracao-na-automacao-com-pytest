```python
import seaborn as sns, pandas as pd, numpy as np, matplotlib.pyplot as plt

df = pd.read_csv('geracao-casos-de-teste-com-llm.csv')

print(df)
```

```python
def plot_performance(df, title):
  plt.figure(figsize=(10, 5))
  plt.gca().set_axisbelow(True)

  sns.barplot(data=df, x='Parâmetro', y='Valor', hue='Modelo', errorbar=('ci', 95), capsize=.1, palette='rocket')

  plt.grid(axis='y', linestyle='--', alpha=0.7, zorder=0)

  # plt.title(title)

  plt.xlabel('')
  plt.ylabel('')
  plt.show()
```

```python
plot_performance(df,'Desempenho Geral')
```

```python
plot_performance(df[df["Teste"] == "Teste 1"],'Desempenho Teste 1')
```

```python
plot_performance(df[df["Teste"] == "Teste 2"],'Desempenho Teste 2')
```

```python
plot_performance(df[df["Teste"] == "Teste 3"],'Desempenho Teste 3')
```

```python
plot_performance(df[df["Teste"] == "Teste 4"],'Desempenho Teste 4')
```
